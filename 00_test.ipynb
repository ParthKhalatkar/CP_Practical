```python
#get marks of students and give a grade based on the marks
# 95 and above - Grade A , 85 and above - Grade B , 75 and above - Grade C , 65 and above - Grade D , 55 and above - Grade E , below 55 - Grade F
marks = int(input("Enter your marks: "))
if marks >= 95:
    print("Grade A")
elif marks >= 85:
    print("Grade B")
elif marks >= 75:
    print("Grade C")
elif marks >= 65:
    print("Grade D")
elif marks >= 55:
    print("Grade E")
else:
    print("Grade F")
```
